# Section 2: Git Fundamentals

## The Three Areas of Git

This is the most important concept to understand. Git organizes your files into three conceptual areas:

```
1. WORKING DIRECTORY  →  2. STAGING AREA  →  3. REPOSITORY
   (Your actual files)    (Changes to save)   (Permanent history)
```

Think of it like preparing a research paper:

1. **Working Directory** = Your Word document as you edit it
2. **Staging Area** = Figures/tables you've decided to include
3. **Repository** = Published versions (draft 1, draft 2, final)

## The Basic Git Workflow

```
1. EDIT files in Working Directory
2. ADD changes to Staging Area (git add)
3. COMMIT staged changes to Repository (git commit)
```

Let's see this in action!

## Step 1: Create a File (Working Directory)

We'll create a simple neuroscience analysis script:

````python
# Create a Python file for calculating firing rates
script_content = """# Firing rate analysis
import numpy as np

def calculate_firing_rate(spike_count, time_window):
    '''Calculate firing rate in Hz'
    
    Parameters:
    spike_count: number of spikes
    time_window: time window in seconds
    
    Returns:
    firing_rate: spikes per second (Hz)
    '''
    firing_rate = spike_count / time_window
    return firing_rate

# Test the function
spikes = 150
time = 10  # seconds
rate = calculate_firing_rate(spikes, time)
print(f"Firing rate: {rate} Hz")
"""

# Write to file
with open('firing_rate.py', 'w') as f:
    f.write(script_content)

print("✓ Created firing_rate.py")
````

Now let's check what Git sees:

````python
# Check repository status
!git status
````

**What Git is telling us:**
- `firing_rate.py` appears in red under "Untracked files"
- Git sees the file in your Working Directory
- But it's NOT yet tracked for version control
- It's NOT staged, and it's NOT committed

**Next step:** Let's stage this file!

## Step 2: Stage the File (git add)

````python
# Add file to staging area
!git add firing_rate.py
````

Let's check the status again:

````python
!git status
````

**What changed:**
- `firing_rate.py` now appears in GREEN under "Changes to be committed"
- The file is now in the Staging Area
- It's ready to be committed to permanent history

**Next step:** Let's commit!

## Step 3: Commit (git commit)

````python
# Commit staged changes with a descriptive message
!git commit -m "Add firing rate calculation function"
````

**What just happened:**
- Git created a permanent snapshot of your staged changes
- The snapshot includes:
  - What changed (the file content)
  - Who made the change (you)
  - When it was made (timestamp)
  - Why it was made (your commit message)

Let's check the status one more time:

````python
!git status
````

**"nothing to commit, working tree clean"** means:
- All changes have been committed
- Your Working Directory matches the Repository
- You're in a clean state!

## The Complete Cycle

Let's do it again to reinforce the workflow. We'll modify the file:

````python
# Add a new function to the file
!echo "" >> firing_rate.py
!echo "# Calculate average firing rate from multiple trials" >> firing_rate.py
!echo "def average_firing_rate(spike_counts, time_window):" >> firing_rate.py
!echo "    rates = [calculate_firing_rate(count, time_window) for count in spike_counts]" >> firing_rate.py
!echo "    return np.mean(rates)" >> firing_rate.py

print("✓ Modified firing_rate.py")
````

````python
# Check status - you'll see the file is modified
!git status
````

Notice `firing_rate.py` is now in RED under "Changes not staged for commit". Git knows the file changed!

Let's see WHAT changed:

````python
# Show differences between working directory and last commit
!git diff firing_rate.py
````

**Reading git diff:**
- Lines starting with `+` are additions (in green)
- Lines starting with `-` are deletions (in red)
- This shows you exactly what changed

Now let's stage and commit this change:

````python
# Stage the changes
!git add firing_rate.py

# Commit with a descriptive message
!git commit -m "Add function to calculate average firing rate across trials"
````

Perfect! You've now completed the full git workflow twice.

## Quick Reference: Essential Commands

```bash
git status          # Check what's changed
git add <file>      # Stage a specific file
git add .           # Stage all changes
git commit -m "msg" # Commit staged changes
git diff            # See unstaged changes
```

# Section 3: Viewing History

One of Git's superpowers is time travel - you can see exactly what happened when.

## The Git Log

The `git log` command shows your commit history:

````python
# View full commit history
!git log
````

**Each commit shows:**
- Commit hash (unique ID): `commit abc123...`
- Author: who made the change
- Date: when it was made
- Message: why it was made

````python
# More compact view - one line per commit
!git log --oneline
````

This is much easier to scan! Each line shows:
- Short hash (first 7 characters)
- Commit message

````python
# View with ASCII graph (useful with branches later)
!git log --oneline --graph --all
````

## View a Specific Commit

You can see details of any commit using its hash:

````python
# First, get the hash of our first commit
!git log --oneline | tail -1
````

Copy the hash from above and paste it in the next cell (replace `HASH_HERE`):

````python
# Replace HASH_HERE with your actual commit hash
# !git show HASH_HERE

# Or use HEAD~1 to see the previous commit
!git show HEAD~1
````

**Understanding HEAD:**
- `HEAD` = your current commit (where you are now)
- `HEAD~1` = one commit before HEAD
- `HEAD~2` = two commits before HEAD

# Section 4: Working with Remote Repositories

So far, everything has been on your local computer. But Git's real power comes from collaboration and backup via **remote repositories**.

## Local vs Remote

- **Local Repository:** On your computer (what we've been using)
- **Remote Repository:** On a server (like GitHub, GitLab, Bitbucket)

Think of it like:
- Local = your lab notebook
- Remote = the lab's shared project folder

## Simulating a Remote

For practice, we'll create a "remote" repository on your own computer. This is safe and lets you practice push/pull operations before using real GitHub.

First, let's create a bare repository (a repository without a working directory):

````python
# Create a "remote" repository
!mkdir -p /tmp/git_remote_repo
!git init --bare /tmp/git_remote_repo
print("✓ Created simulated remote repository")
````

Now let's connect our repository to this "remote":

````python
# Add remote repository
!git remote add origin /tmp/git_remote_repo

# Verify it was added
!git remote -v
````

**What this shows:**
- `origin` is the conventional name for your main remote
- `fetch` is for downloading from the remote
- `push` is for uploading to the remote

## Pushing to Remote (git push)

Let's send our commits to the remote:

````python
# Push commits to remote
# -u sets "upstream" - links your local branch to remote branch
!git push -u origin main
````

Your commits are now backed up to the remote!

## Making More Changes

Let's make another change and push it:

````python
# Add analysis notes
with open('notes.md', 'w') as f:
    f.write("""# Analysis Notes

## 2026-01-28
- Implemented firing rate calculations
- Added average firing rate function
- Created spike data analysis script
""")

!git add notes.md
!git commit -m "Add analysis notes"
print("✓ Created new commit locally")
````

## The GitHub Connection

What we just practiced is exactly what you do with GitHub:

1. Create a repository on GitHub (website)
2. Connect your local repository to it: `git remote add origin <GitHub-URL>`
3. Push your commits: `git push -u origin main`
4. Future pushes: just `git push`

The process is identical - the only difference is the remote URL points to GitHub instead of `/tmp`.

## Pulling from Remote (git pull)

When collaborating, others might add commits to the remote. You download them with `git pull`:

````python
# Pull any new commits from remote
!git pull
````

"Already up to date" means there are no new commits on the remote.

## Quick Reference: Remote Commands

```bash

git remote -v                # List remote connections
git push -u origin main      # Push commits (first time)

git pull                     # Download new commits from remote
```

# Section 5: Common Scenarios & Best Practices

## Undoing Mistakes

Everyone makes mistakes. Git helps you recover!

### Scenario 1: Unstage a File

You staged something by accident:

````python
# Create a temporary file
!echo "temporary stuff" > temp.txt

# Oops, staged it by accident
!git add temp.txt

# Check status
!git status
````

To unstage:

````python
# Unstage the file
!git reset temp.txt

# Check status - back to untracked
!git status
````

The file still exists, but it's no longer staged.

### Scenario 2: Discard Changes to a File

You modified a file but want to throw away the changes:

````python
# Modify a file
!echo "\n# Bad change" >> firing_rate.py

# Check status
!git status
````

To discard the changes:

````python
# Restore file to last committed version
!git restore firing_rate.py

# Check status - clean again
!git status
````

⚠️ **WARNING:** `git restore` permanently deletes your changes. Use carefully!

## The .gitignore File

Some files shouldn't be tracked by Git:
- Large data files
- Temporary files
- Generated outputs
- Sensitive information (passwords, API keys)

Create a `.gitignore` file to tell Git what to ignore:

````python
# Create a neuroscience-specific .gitignore
gitignore_content = """# Python
__pycache__/
*.pyc
*.pyo
.ipynb_checkpoints/

# Data files (usually too large for git)
*.npy
*.mat
*.h5
data/raw/
data/processed/

# Analysis outputs
figures/
results/
*.png
*.jpg
*.pdf

# Environment
.env
venv/
env/

# IDE
.vscode/
.idea/
*.swp

# OS
.DS_Store
Thumbs.db

# Temporary files
temp.txt
*.tmp
"""

with open('.gitignore', 'w') as f:
    f.write(gitignore_content)

print("✓ Created .gitignore")
````

````python
# Check status - temp.txt is now ignored!
!git status
````

Notice `temp.txt` no longer appears as untracked. It's being ignored.

Let's commit the .gitignore:

````python
!git add .gitignore
!git commit -m "Add gitignore for neuroscience project"
````

## Writing Good Commit Messages

Your future self (and collaborators) will thank you for clear commit messages.

**Good commit messages:**
- ✅ "Add spike sorting function with threshold detection"
- ✅ "Fix bug in firing rate calculation for zero time windows"
- ✅ "Update README with installation instructions"

**Bad commit messages:**
- ❌ "update"
- ❌ "fix stuff"
- ❌ "asdf"
- ❌ "final version" (there's always another version!)

**Format:**
```
Start with a verb (Add, Fix, Update, Remove, Refactor)
Be specific but concise (50 characters or less)
Explain WHAT and WHY, not HOW (code shows how)
```

## Best Practices for Research Code

1. **Commit early and often**
   - Commit after each logical change
   - Don't wait until "everything is perfect"

2. **One commit = one logical change**
   - Don't mix unrelated changes
   - Makes it easier to understand history

3. **Don't commit generated data**
   - Use .gitignore for large data files
   - Commit the CODE that generates data, not the data itself

4. **Commit before risky operations**
   - Before major refactoring
   - Before trying something experimental
   - Gives you a safe point to return to

5. **Keep sensitive info out**
   - Never commit passwords, API keys, or private data
   - Use .env files and add them to .gitignore
